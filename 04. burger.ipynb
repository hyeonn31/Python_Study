```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
sales_df = pd.DataFrame({
    'quarter': ['1Q', '2Q', '3Q', '4Q'],
    'revenue': [1360, 2650, 2070, 4150],
    'cost': [1240, 1970, 1750, 2760]
})

sales_df
```

```python
plt.plot(sales_df['quarter'], sales_df['revenue'])
plt.show()
```

```python
sales_df.plot(x='quarter', y= 'revenue')
plt.show()
```

```python
sales_df.plot(x='quarter', y= 'revenue',kind='bar')
plt.show()
```

```python
sales_df.plot(y= 'revenue',kind='pie',labels=['Q1','Q2','Q3','Q4'], ylabel='revenue')
plt.show()
```
