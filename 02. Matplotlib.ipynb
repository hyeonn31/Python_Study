```python
import numpy as np                 # 숫자 배열(데이터)을 다루는 라이브러리
import matplotlib.pyplot as plt    # 그래프를 그리는 라이브러리 (plt 라는 별명으로 줄여 쓰는 것이 관례)
```

```python
# 연도별 주가 데이터 (같은 위치의 값끼리 짝이 됨: 2011년 → 14.46)
year_array = np.array([2011, 2012, 2013, 2014, 2015, 2016, 2017, 2018, 2019, 2020])

stock_array = np.array([
    14.46, 19.01, 20.04, 27.59, 26.32,
    28.96, 42.31, 39.44, 73.41, 132.69
])
```

```python
# 선 그래프(line plot) : 시간에 따른 변화(추세)를 볼 때 사용
plt.plot(year_array , stock_array)  # plt.plot(x값, y값) : 점들을 순서대로 선으로 이어 줌
plt.show()                          # 완성된 그래프를 화면에 출력
```

```python
# 후보 이름과 득표수 데이터
name_array = np.array(['mark','dongwook','hyojune','sowon','taeho'])
vote_array = np.array([5,10,6,8,3])
```

```python
# 막대 그래프(bar plot) : 항목별 크기를 비교할 때 사용
plt.bar(name_array, vote_array)  # plt.bar(x축 항목, 막대 높이)
plt.show()
```

```python
# 30명의 키(cm)와 몸무게(kg) 데이터 (같은 위치의 값이 같은 사람)
height_array = np.array([
    165, 164, 155, 151, 157, 162, 155, 157, 165, 162,
    165, 167, 167, 183, 180, 184, 177, 178, 175, 181,
    172, 173, 169, 172, 177, 178, 185, 186, 190, 187
])

weight_array = np.array([
    62, 59, 57, 55, 60, 58, 51, 56, 68, 64,
    57, 58, 64, 79, 73, 76, 61, 65, 83, 80,
    67, 82, 88, 62, 61, 79, 81, 68, 83, 80
])
```

```python
# plt.scatter(x값, y값) : 점 하나가 한 사람 (가로 위치 = 키, 세로 위치 = 몸무게)
plt.scatter(height_array,weight_array)  #산점도(scatter plot) : 흩뿌리다 (두 변수의 상관관계)
plt.show()
```

```python
plt.scatter(height_array,weight_array)  #산점도(scatter plot) : 흩뿌리다 (두 변수의 상관관계)
plt.title('Height and Weight')  # 그래프 제목
plt.show()
```

```python
plt.scatter(height_array,weight_array)  #산점도(scatter plot) : 흩뿌리다 (두 변수의 상관관계)
plt.title('Height and Weight')
plt.xlabel('Heigth (cm)')  # x축 이름
plt.ylabel('Weight (kg)')  # y축 이름
plt.show()
```

```python
# c : 점의 색 (color). 'red' 처럼 색 이름으로 지정
plt.scatter(height_array,weight_array,c = 'red')  #산점도(scatter plot) : 흩뿌리다 (두 변수의 상관관계)
plt.title('Height and Weight')
plt.xlabel('Heigth (cm)')
plt.ylabel('Weight (kg)')
plt.show()
```

```python
# 다른 색 이름도 가능 ('green', 'blue', 'orange' 등)
plt.scatter(height_array,weight_array,c='green')  #산점도(scatter plot) : 흩뿌리다 (두 변수의 상관관계)
plt.title('Height and Weight')
plt.xlabel('Heigth (cm)')
plt.ylabel('Weight (kg)')
plt.show()
```

```python
# 색상 코드(hex)로 원하는 색을 정확히 지정할 수도 있음 ('#' + 16진수 6자리)
plt.scatter(height_array,weight_array,c = '#6500c3')  #산점도(scatter plot) : 흩뿌리다 (두 변수의 상관관계)
plt.title('Height and Weight')
plt.xlabel('Heigth (cm)')
plt.ylabel('Weight (kg)')
plt.show()
```

```python
# marker : 점의 모양. '+' 는 더하기 모양 (기본값은 원 'o')
plt.scatter(height_array,weight_array,marker='+',c='red')  #산점도(scatter plot) : 흩뿌리다 (두 변수의 상관관계)
plt.title('Height and Weight')
plt.xlabel('Heigth (cm)')
plt.ylabel('Weight (kg)')
plt.show()
```

```python
# marker='s' : 네모(square) 모양
plt.scatter(height_array,weight_array, marker='s')  #산점도(scatter plot) : 흩뿌리다 (두 변수의 상관관계)
plt.title('Height and Weight')
plt.xlabel('Heigth (cm)')
plt.ylabel('Weight (kg)')
plt.show()
```

```python
# plt.figure(figsize=(가로, 세로)) : 이 그래프 하나의 크기만 지정 (단위: 인치)
# 반드시 그래프를 그리는 코드(scatter)보다 먼저 적어야 함
plt.figure(figsize=(10, 4))
plt.scatter(height_array, weight_array)
plt.title('Height and Weight')
plt.xlabel('Height (cm)')
plt.ylabel('Weight (kg)')
plt.show()
```

```python
# plt.rcParams : matplotlib 의 기본 설정값 모음
# 여기서 크기를 바꾸면 이후에 그리는 모든 그래프에 적용됨 (커널 재시작 전까지 유지)
plt.rcParams['figure.figsize'] = (5, 5)
plt.scatter(height_array, weight_array)
plt.title('Scatter Plot')
plt.xlabel('X Label')
plt.ylabel('Y Label')
plt.show()
```

## 연봉 데이터 시각화

### 실습 설명
어떤 회사 직원들의 나이와 연봉 정보를 시각화해 봅시다. 아래 요구 사항을 모두 지켜서 실습 결과와 동일한 그래프를 그려 보세요. 오타와 띄어쓰기에 주의해 주세요!

- 그래프 제목: Age and Salary
- x축 제목: Age
- y축 제목: Salary($)
- 점의 색깔: 빨간색(red)
- 점의 모양: 네모
- 실습 결과


```python
import numpy as np
import matplotlib.pyplot as plt

# 직원 30명의 나이와 연봉 데이터
age_array = np.array([
	27, 25, 29, 22, 26, 28, 40, 42, 34, 31,
	37, 36, 40, 32, 41, 47, 52, 49, 51, 50,
	49, 46, 47, 46, 47, 45, 57, 58, 60, 56
])

salary_array = np.array([
	44009, 83450, 95025, 63748, 78652,
	58977, 114733, 126353, 121904, 125083,
	143739, 140754, 144346, 114899, 125806,
	180252, 153374, 171057, 161305, 193786,
	167238, 150157, 179843, 195326, 165849,
	164887, 131372, 135876, 113796, 143050
])

plt.rcParams['figure.figsize'] = (6, 4)                    # 그래프 기본 크기 (가로 6, 세로 4)
plt.scatter(age_array, salary_array, marker='s',c='red')   # x: 나이, y: 연봉 / 네모 모양, 빨간색
plt.title('Age and Salary')                                # 그래프 제목
plt.xlabel('Age')                                          # x축 이름
plt.ylabel('Salary($)')                                    # y축 이름
plt.show()
```

## 그래프에 한글로 된 텍스트 넣기

```python
# 기본 글꼴에는 한글이 없어서 한글이 네모(□)로 깨짐 → 한글 글꼴로 바꿔 줘야 함
# 주의: 그래프를 그리는 코드(scatter)보다 먼저 적어야 적용됨
plt.rc('font',family = 'Malgun Gothic' )  # 한글 적용 (Malgun Gothic = 윈도우의 맑은 고딕)
plt.scatter(height_array, weight_array, c='#6500c3', marker='s')
plt.title('키와 몸무게')
plt.xlabel('키(cm)')
plt.ylabel('몸무게(kg)')
plt.show()
```
