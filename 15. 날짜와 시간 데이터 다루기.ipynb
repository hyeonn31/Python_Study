```python
# 이런 형식의 데이터를 불러올 때에는 dayfirst라는 파라미터를 True로 설정
# 이건 말 그대로 일자 값이 앞에 온다는 것을 의미 원래 기본값은 False로 되어 있어서,
# 따로 설정해 주지 않으면 월/일/연도 순서로 값을 인식
# 물론 따로 이 파라미터를 설정하지 않아도 20/01/2015 이런 식으로 
# 가장 앞에 있는 값이 월이 아니라 일이라는 것이 명확하면 pandas가 알아서 
# 일/월/연도 형식으로 판단하여 데이터를 불러옴. (20이라는 월은 존재하지 않기 때문)
# 하지만, 10/01/2015처럼 월/일/연도인지, 일/월/연도인지 명확하지 않은 경우도 많기 때문에 
# 따로 파라미터 값을 명확하게 설정해 주는 것이 좋음.


order_df = pd.read_csv('data/order3.csv', parse_dates=['order_time', 'shipping_time'], dayfirst=True)
order_df.head()
```

## 데이터 타입을 변경할 때 형식 지정하기 

```python
# 년도가 2자리일 때 
order_df = pd.read_csv('data/order4.csv')
order_df.head()
```

```python
order_df['order_time'] = pd.to_datetime(order_df['order_time'], yearfirst=True)
order_df['shipping_time'] = pd.to_datetime(order_df['shipping_time'], yearfirst=True)

order_df.head()
```

```python
order_df = pd.read_csv('data/order5.csv')
order_df.head()
```

```python
order_df['order_time'] = pd.to_datetime(order_df['order_time'], format='%Y년 %m월 %d일 %H시 %M분 %S초')
order_df['shipping_time'] = pd.to_datetime(order_df['shipping_time'], format='%Y년 %m월 %d일 %H시 %M분 %S초')
order_df.head()
```

## datetime을 원하는 형태로 바꿔서 표현하기

 반대로 datetime 값을 원하는 형태의 문자열로 바꿔서 표현하는 것도 가능  
예를 들어 2017-10-31 16:37:00이라는 값을 -> 31 October 2017, 04:37PM

```python
order_df['order_time'] = order_df['order_time'].dt.strftime('%d %b %Y, %I:%M %p')
order_df['shipping_time'] = order_df['shipping_time'].dt.strftime('%d %b %Y, %I:%M %p')

order_df.head()
```

```python
import pandas as pd

order_df = pd.read_csv('data/order.csv')
order_df
```

```python
order_df.dtypes
```

```python
order_df= order_df.set_index('order_time').sort_index()
```

```python
order_df['shipping_time'] - order_df.index
```

```python
order_df['days_for_shipping(actual)'] = order_df['shipping_time'] - order_df.index
```
