## 결측값 찾기

```python
import pandas as pd

airbnb_df = pd.read_csv('data/airbnb.csv')
airbnb_df
```

```python
airbnb_df.info()
```

```python
airbnb_df.isna()   # 결측값 여부
```

```python
airbnb_df.isna().sum()  # rating에 결측값이 13개 있음
```

```python
airbnb_df.isna().any(axis=1)  #각 row별로 결측값의 존재 여부를 알 수 있음
```

```python
airbnb_df[airbnb_df.isna().any(axis=1)]  # 결측값이 있는 데이터만 뽑아서
```

```python
airbnb_df.dropna()  # 결측값 삭제
```

```python
airbnb_df['rating'].mean().round(0)
```

```python
airbnb_mean = airbnb_df['rating'].mean().round(0)
```

```python
airbnb_df['rating'].fillna(airbnb_mean)
```

```python
# 결측값이 있던 부분에 잘 채워졌는지 mask를 사용해서 다시 확인
mask = airbnb_df['rating'].isna()
airbnb_df['rating'].fillna(airbnb_mean)[mask]
```

```python
airbnb_df['rating'] = airbnb_df['rating'].fillna(airbnb_mean)
```

## 이상점 찾기 (Outlier)

```python
q1 = airbnb_df['price'].quantile(0.25)
q3 = airbnb_df['price'].quantile(0.75)
```

```python
iqr = q3 - q1
```

```python
lower_limit = q1 - 1.5*iqr
upper_limit = q3 + 1.5*iqr
```

```python
airbnb_df['price'] < lower_limit
```

```python
airbnb_df[airbnb_df['price'] < lower_limit]
```

```python
airbnb_df[airbnb_df['price'] > upper_limit]
```

```python
airbnb_df[(airbnb_df['price'] < lower_limit )| (airbnb_df['price'] > upper_limit)]
```

### 이상점 처리하기

```python
# 정상범위
condition1 = airbnb_df['price'] >= lower_limit
condition2 = airbnb_df['price'] <= upper_limit
```

```python
airbnb_df[condition1 & condition2]
```
