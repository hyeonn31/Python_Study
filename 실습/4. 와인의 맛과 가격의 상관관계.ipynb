## 비싼 와인이 더 맛있을까?

실습 설명  
윤수는 동욱이의 집들이 선물로 와인을 준비했다.   
백화점에 가서 와인을 고르는데 왠지 가격이 비쌀수록 맛이 좋을 것 같아서 비싼 와인을 샀는데    
다행히 동욱이는 와인을 아주 마음에 들어 했지만, 윤수는 정말로 와인 가격이 비쌀수록 맛이 좋다고 할 수 있는지 궁금해졌다.   
그래서 다양한 가격대의 와인 데이터를 직접 수집했다.   
와인 가격과 평점의 관계를 볼 수 있는 그래프를 그려보자.  


```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

wine_df = pd.read_csv('../data/wine.csv')
sns.set_theme(rc={'figure.figsize': (6, 6)}, style='white')

wine_df
```

```python
sns.regplot(data=wine_df, x='price', y='points')
plt.show()
```
