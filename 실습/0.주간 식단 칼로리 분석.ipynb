## 데이터

```python
# 1인분 기준 용량 (g)
food_grams = {
    "현미밥": 210,
    "닭가슴살": 150,
    "계란": 60,
    "고구마": 130,
    "샐러드": 200,
    "바나나": 120,
    "두부": 150,
    "오트밀": 150,
    "삼겹살": 200,
}

# 1g당 칼로리 (kcal)
food_calories_per_gram = {
    "현미밥": 1.43,
    "닭가슴살": 1.33,
    "계란": 1.35,
    "고구마": 1.31,
    "샐러드": 0.75,
    "바나나": 0.84,
    "두부": 1.00,
    "오트밀": 2.33,
    "삼겹살": 3.75,
}

weekly_diet = {
    "월": ["오트밀", "바나나", "현미밥", "닭가슴살", "샐러드", "현미밥", "두부", "계란"],
    "화": ["오트밀", "계란", "현미밥", "닭가슴살", "고구마", "두부", "샐러드", "바나나"],
    "수": ["삼겹살", "현미밥", "닭가슴살", "샐러드", "두부", "고구마", "계란"],
    "목": ["오트밀", "바나나", "현미밥", "닭가슴살", "두부", "현미밥", "계란", "샐러드"],
    "금": ["오트밀", "계란", "현미밥", "닭가슴살", "샐러드", "고구마", "두부", "바나나"],
    "토": ["현미밥", "계란", "바나나", "삼겹살", "현미밥", "샐러드", "오트밀", "두부"],
    "일": ["오트밀", "바나나", "현미밥", "닭가슴살", "고구마", "현미밥", "계란", "샐러드"],
}
```

## 1. 음식별 1인분 칼로리

```python
# 1. 음식별 1인분 칼로리를 계산하는 함수
def calculate_food_calories(food_grams, food_calories_per_gram):
    # food_grams를 순회하면서 음식 이름을 하나씩 가져오세요
    # 해당 음식의 용량(food_grams)과 1g당 칼로리(food_calories_per_gram)를 가져오세요
    # 두 값을 곱해 칼로리를 계산하고 int()로 변환하세요
    # 계산한 값을 food_calories 딕셔너리에 저장하세요
    # 여기에 코드를 작성하세요
    food_calories = {}

    for food in food_grams:   #음식 이름을 하나씩 꺼냄
        grams = food_grams[food]  # 1인분 용량(g)
        per_gram = food_calories_per_gram[food]    #1g당 칼로리
        food_calories[food] = int(grams * per_gram)


    return food_calories
```

```python
food_calories = calculate_food_calories(food_grams, food_calories_per_gram)
food_calories
```

## 2. 요일별 총 칼로리

```python
# 2. 요일별 총 칼로리를 계산하는 함수
def calculate_daily_calories(weekly_diet, food_calories):
    # weekly_diet를 순회하면서 각 요일마다 총 칼로리를 계산하세요
    # 각 음식의 칼로리는 food_calories에서 가져오고,
    # 계산한 결과를 daily_calories에 저장하세요
    # 여기에 코드를 작성하세요
    daily_calories = {}

    for i in weekly_diet: # "월", "화", ...
        foods = weekly_diet[i] # 그날 먹은 음식 리스트                       # ② 그날 음식 리스트
        total = 0
        for j in foods: # "오트밀", "바나나", ...
            total += food_calories[j] # 음식별 1인분 칼로리를 누적
        daily_calories[i] = total


    return daily_calories
```

```python
daily_calories = calculate_daily_calories(weekly_diet, food_calories)
daily_calories
```

## 3. 목표 칼로리 달성 여부

```python
# 3. 목표 칼로리 달성 여부를 판단하는 함수
def evaluate_daily_calories(daily_calories, daily_goal=1800):
    # 요일별 총 칼로리를 순회하면서 목표 칼로리와 비교하세요
    # 목표 칼로리를 상단에 출력하세요
    # 차이가 음수이면 "목표 달성", 0이면 "완벽 달성", 양수이면 "목표 초과"를 출력하세요
    # 여기에 코드를 작성하세요
    print(f"목표 칼로리: {daily_goal}kcal")
    for i in daily_calories:
        if daily_calories[i] - daily_goal < 0 :
            print(f"{i} -> 목표 달성")
        elif daily_calories[i] - daily_goal == 0:
            print(f"{i} -> 완벽 달성") 
        else :
            print(f"{i} -> 목표 초과")






```

```python
evaluate_daily_calories(daily_calories)
```

## 4. 주간 통계

```python
# 4. 주간 통계를 계산하는 함수
def calculate_weekly_stats(daily_calories):
    # 첫 번째 요일을 기준으로 최대/최소 값을 초기화합니다
    # total_sum을 0으로 초기화합니다
    days = list(daily_calories.keys())
    max_day = min_day = days[0]
    max_value = min_value = daily_calories[days[0]]
    total_sum = 0
    
    # daily_calories를 순회하면서
    # total_sum에 값을 누적하고,
    # max_value, min_value를 비교해 갱신하세요
    # 여기에 코드를 작성하세요

    for i in daily_calories:       
        today = daily_calories[i]  # 오늘 칼로리 
        total_sum += today         # 합계 누적

        if today > max_value :     # 오늘이 지금까지 1등보다 크면
            max_value = today
            max_day = i

        if today < min_value :     # 오늘이 지금까지 꼴등보다 작으면
            min_value = today
            min_day = i

    average = total_sum / len(days) 


    return round(average), max_day, min_day
```

```python
calculate_weekly_stats(daily_calories)
```

## 5. 목표 달성/초과 날짜 분류

```python
# 5. 목표 달성/초과 날짜를 분류하는 함수
def categorize_days(daily_calories, daily_goal=1800):
    # 목표 칼로리 이하인 날은 success 리스트에,
    # 목표 칼로리를 초과한 날은 excess 리스트에 담아 반환하세요
    # 리스트 컴프리헨션을 사용해 보세요
    # 여기에 코드를 작성하세요

    success = [ i for i in daily_calories if daily_calories[i] <= daily_goal]
    excess = [ j for j in daily_calories if daily_calories[j] > daily_goal]

    return success, excess
```

```python
categorize_days(daily_calories)
```

## 6. 전체 분석

```python
# 6. 전체 분석을 수행하는 함수
def analyze_weekly_diet(weekly_diet):
    # 1) 음식별 칼로리를 계산하세요.
    food_calories = calculate_food_calories(food_grams, food_calories_per_gram)


    # 2) 요일별 총 칼로리를 계산하세요.
    daily_calories = calculate_daily_calories(weekly_diet, food_calories)


    # 3) 음식별 1인분 칼로리와 요일별 총 섭취 칼로리를 출력하세요
    print("[음식별 1인분 칼로리]")
    for food in food_calories:
        print(f"{food}: {food_calories[food]}kcal")

    print("\n[요일별 총 섭취 칼로리]")
    for day in daily_calories:
        print(f"{day}: {daily_calories[day]}kcal")

    # 4) 목표 칼로리 비교 결과를 출력하세요
    print()
    evaluate_daily_calories(daily_calories)


    # 5) 주간 통계를 계산하고 출력하세요
    average, max_day, min_day = calculate_weekly_stats(daily_calories) # 4번 함수의 마지막 줄은 return average, max_day, min_day였음. 값 3개를 한 묶음(튜플)으로 돌려줌.
    print("\n[주간 통계]")                                              # 변수를 순서대로 하나씩 나눠 담음. (= 언패킹) 
    print(f"평균 섭취 칼로리: {average:.1f}kcal")
    print(f"최대 섭취 요일: {max_day}")
    print(f"최소 섭취 요일: {min_day}")


    # 6) 목표 달성/초과 날짜를 분류해 출력하세요
    success, excess = categorize_days(daily_calories)
    print("\n[목표 달성/초과 분류]")
    print("목표 달성:", success)
    print("목표 초과:", excess)



```

```python
analyze_weekly_diet(weekly_diet)
```
